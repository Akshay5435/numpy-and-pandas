```python
# ==========================================
# NumPy Section
# ==========================================

#Create a 1D NumPy Array:
import numpy as np

temperatures_w1 = np.array([22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9])
```

```python
# ------------------------------------------------------------
# Inspection and Properties
# ------------------------------------------------------------
#Inspect the shape, data type, and the number of elements in the array.Inspect the shape of the temperatures_w1 array

#shape of the array
print("Shape:",temperatures_w1.shape)
#data type of the array
print("DataType:",temperatures_w1.dtype)
#size of the array
print("No.of elements:",temperatures_w1.size)
```

```python
# ------------------------------------------------------------
# Array Operations
# ------------------------------------------------------------
# ● Convert the temperatures_w1 from Celsius to Fahrenheit using the formula: Fahrenheit = (Celsius * 9/5) + 32

fahrenheit = (temperatures_w1 * 9/5) + 32

print("Temperatures in Celsius:",temperatures_w1)
print("Temperatures in Fahrenheit:",fahrenheit)
```

```python
# ● Find the maximum, minimum, and mean temperatures for the week.
#Maximum temperature
maximum_temperature = np.max(temperatures_w1)
print("Maximum temperature:",maximum_temperature,"°C")
#Minimum temperature
minimum_temperature = np.min(temperatures_w1)
print("Minimum temperature:",minimum_temperature, "°C")
#Mean temperature
mean_temperature = np.mean(temperatures_w1)
print("Mean temperature:",mean_temperature, "°C")
```

```python
# ------------------------------------------------------------
# 4. Array Slicing and Indexing:
# ------------------------------------------------------------
# ● Extract the temperatures for the first three days of the week.

# First three days
first_three_days = temperatures_w1[:3]
print("First Three Days:",first_three_days)
```

```python
# ● Extract the temperatures for the weekend (last two days).

# Last two days
weekend_temperatures = temperatures_w1[-2:]
print("Weekend Temperatures:",weekend_temperatures)
```

```python
# ● Extract the temperatures for the middle three days of the week

# Midddle three days
middle_three_days = temperatures_w1[2:5]
print("Middle Three Days:",middle_three_days)
```

```python
# ==========================================
#      NumPy 2D Array
# ==========================================
#5. Create a 2D Array:

#★ Row 1: [22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9] (Week 1)
#★ Row 2: [19.2, 22.5, 21.3, 24.0, 23.5, 22.8, 20.1] (Week 2)

temperatures = np.array([
    [22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9],
    [19.2, 22.5, 21.3, 24.0, 23.5, 22.8, 20.1]
 ])
```

```python
# ------------------------------------------------------------
# 6. Inspect and Slice the 2D Array:
# ------------------------------------------------------------

# ● Inspect and print the shape, data type, and total number of elements in the temperatures array.
print("Shape:", temperatures.shape)
print("DataType:", temperatures.dtype)
print("Total no.of elements:", temperatures.size)

# ● Extract the temperatures for each week and the weekends (last two days) for both weeks.
week1 = temperatures[0]
print("Week 1 Temperatures:",week1)  #each week

week2 = temperatures[1]
print("Week 2 Temperatures:",week2)   #each week

week1_weekend = temperatures[0, -2:]
print("Week 1 Weekend Temperatures:", week1_weekend)
week2_weekend = temperatures[1, -2:]
print("Week 2 Weekend Temperatures:", week2_weekend)
```

```python
# ==========================================
# #Pandas Series:
# ==========================================

# ------------------------------------------------------------
#1. Creating Pandas Series:
# ------------------------------------------------------------
# ➢ Marks: 95, 92, 89, 85, 80
# ➢ Custom Indices: 'Rank1', 'Rank2', 'Rank3', 'Rank4', 'Rank5'

import pandas as pd

marks = pd.Series([95, 92, 89, 85, 80],
                  index=['Rank1','Rank2','Rank3','Rank4','Rank5'])
print("Marks Series:",marks)
```

```python
# ------------------------------------------------------------
# 2. Indexing and Slicing:
# ------------------------------------------------------------
# ● Use the integer index position to access the mark of the 1st rank student.
first_rank_mark = marks.iloc[0]
print("Mark of 1st Rank Student:", first_rank_mark)
```

```python
# ● Use the loc accessor to retrieve the marks of the top 3 ranks by specifying their index labels.
top_three_marks = marks.loc[['Rank1', 'Rank2', 'Rank3']]
print("Top 3 Rank Marks:",top_three_marks)
```

```python
# ● Use the iloc accessor to retrieve the mark of the 3rd rank student.
third_rank_mark = marks.iloc[2]
print("Mark of 3rd Rank Student:", third_rank_mark)
```

```python
# ● Apply a boolean mask to filter the Series and retrieve the ranks where the marks are greater than 90.
marks_above_90 = marks[marks > 90]
print("Ranks with Marks Greater Than 90:",marks_above_90)
```

```python
# ------------------------------------------------------------
# 3. Manipulating Series:
# ------------------------------------------------------------
# ● Modify the mark of the student with the 1st rank to 100.
marks.loc['Rank1'] = 100
print("Updated Marks Series:",marks)
```

```python
# ● Remove the entry corresponding to the last rank from the Series.
marks = marks.drop('Rank5')
print("Marks Series After Removing Rank5:",marks)
```

```python
# ● Compute the CGPA by dividing each mark in the Series by 10.
cgpa = marks / 10
print("CGPA:",cgpa)
```

```python
# ==========================================
#Pandas DataFrame
# ==========================================

# 1. Creating Pandas DataFrame:
# Create transactions DataFrame

transactions = pd.DataFrame({
    'TransactionID': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],

    'ProductCategory': ['Electronics','Clothing','Electronics','Furniture','Clothing',
        'Electronics','Furniture','Clothing','Furniture','Electronics'],

    'Region': ['North','South','North','East','West',
        'North','East','West','South','North'],

    'Amount': [200,150,300,450,200,
        250,300,180,350,400]
})

print("Transactions DataFrame:",transactions)
```

```python
# ------------------------------------------------------------
#2. Data Exploration:
# ------------------------------------------------------------
# ● Display the transactions DataFrame and its basic information, including head, tail, shape, column names, and data types.

print("Complete Transactions DataFrame:",transactions)  #Display the Complete DataFrame

print("First 5 Rows:",transactions.head())              #Display the First 5 Rows

print("Last 5 Rows:",transactions.tail())               #Display the last 5 Rows

print("Shape of DataFrame:",transactions.shape)         #Display the Shape

print("Column Names:",transactions.columns)             #Display the Column names

print("Data Types:",transactions.dtypes)                #Display the Data Types
```

```python
# ● Display only the 'ProductCategory' and 'Amount' columns.
print("Product Category and Amount:",transactions[['ProductCategory', 'Amount']])
```

```python
# ● Retrieve the last 3 columns of the DataFrame using iloc or loc.
print("Last 3 Columns:",transactions.iloc[:, -3:])
```

```python
# ● Filter rows where the 'Region' is 'North' and 'Amount' is greater than 200.
north_high_amount = transactions[
    (transactions['Region'] == 'North') &
    (transactions['Amount'] > 200)]
print("North Region with Amount Greater Than 200:",north_high_amount)
```

```python
# ● Find the value counts for the 'ProductCategory' column.
product_category_counts = transactions['ProductCategory'].value_counts()
print("Product Category Value Counts:",product_category_counts)
```

```python
# ● Find the unique values in the 'Region' column.
unique_regions = transactions['Region'].unique()
print("Unique Regions:",unique_regions)
```

```python
# ● Group by 'Region' and find the mean amount for each region.
mean_amount_by_region = transactions.groupby('Region')['Amount'].mean()
print("Mean Amount by Region:",mean_amount_by_region)
```

```python
# ------------------------------------------------------------
#3. Manipulating the DataFrame
# ------------------------------------------------------------
# ● Modify the 'Amount' for the transaction with TransactionID 102 to 165.
# Modify Amount for TransactionID 102

transactions.loc[transactions['TransactionID'] == 102,'Amount'] = 165

print("DataFrame After Updating TransactionID 102:",transactions)
```

```python
# ● Add a new column 'Discount' by calculating 10% of the 'Amount'.
# Add Discount column

transactions['Discount'] = transactions['Amount'] * 0.10

print("DataFrame After Adding Discount:",transactions)
```

```python
# ● Remove the row with TransactionID 109.

transactions = transactions[transactions['TransactionID'] != 109]

print("DataFrame After Removing TransactionID 109:",transactions)
```

```python
# ● Delete the 'Discount' column from the DataFrame.

transactions = transactions.drop(columns=['Discount'])

print("Final Transactions DataFrame:",transactions)
```
